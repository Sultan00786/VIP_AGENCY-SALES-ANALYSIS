```python
def remove_tail_nulls(dfs, df_name, key_col_name):
    temp_df = dfs.__dict__[df_name]
    dfs.__dict__[df_name] = temp_df[~pd.isna(temp_df[key_col_name])]
    # display(temp_df[~pd.isna(temp_df[key_col_name])])
```

```python
remove_tail_nulls(
    dfs=dfs, 
    df_name='collection_summary', 
    key_col_name='Bill No'
)
```
